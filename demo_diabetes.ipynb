```python
from DPClustX import ClusteringExplainer
import pandas as pd
import warnings
warnings.filterwarnings("ignore", category=UserWarning)  

clustered_df = pd.read_csv('datasets/cdc_diabetes.csv')
```

```python
clustered_df['cluster'] = clustered_df['cluster'] + 1
```

```python
explainer = ClusteringExplainer(clustered_df, mode='admin')
explainer.explain()
```

```python
exclude = {
    3: 'Income'
}
explainer.explain(exclude=exclude)
```

```python
explainer = ClusteringExplainer(clustered_df, mode='user')
explainer.explain()
```

```python
explainer.explain(eps_topcomb=0.001, eps_candlist=0.01)
```
